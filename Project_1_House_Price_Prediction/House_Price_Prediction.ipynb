**Data Preprocessing and Feature Selection**

```python

# Separate features and target variable

X = df.drop('price', axis=1)
y = df['price']

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nInput Features:")
print(X.columns.tolist())

print("\nTarget Variable: price")
```

```python

# Identify numerical and categorical columns

numerical_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object']
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Numerical Features:", len(numerical_features))
print("Total Categorical Features:", len(categorical_features))

```

```python

from sklearn.model_selection import train_test_split

# Split the dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display dataset sizes

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python

from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
```

```python

# Numerical features
numerical_features = [
    'area',
    'bedrooms',
    'bathrooms',
    'stories',
    'parking'
]

# Nominal categorical features
nominal_features = [
    'mainroad',
    'guestroom',
    'basement',
    'hotwaterheating',
    'airconditioning',
    'prefarea'
]

# Ordinal categorical feature
ordinal_features = ['furnishingstatus']

# Define the natural order
furnishing_order = [
    ['unfurnished', 'semi-furnished', 'furnished']
]
```

```python

# Create preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        # Keep numerical features unchanged
        ('numerical', 'passthrough', numerical_features),

        # Apply One-Hot Encoding
        ('nominal',
         OneHotEncoder(drop='first', handle_unknown='ignore'),
         nominal_features),

        # Apply Ordinal Encoding
        ('ordinal',
         OrdinalEncoder(categories=furnishing_order),
         ordinal_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python

# Fit preprocessing on training data

X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data

X_test_processed = preprocessor.transform(X_test)

print("Preprocessing Completed Successfully!")

print("\nTraining Data Shape:", X_train_processed.shape)
print("Testing Data Shape:", X_test_processed.shape)
```

```python

# Get processed feature names

processed_feature_names = preprocessor.get_feature_names_out()

print("Processed Feature Names:\n")

for feature in processed_feature_names:
    print(feature)

print("\nTotal Processed Features:", len(processed_feature_names))

```

**Train the Linear Regression Model**

```python
# Import required libraries

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# Create the complete machine learning pipeline

model_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

# Train the model

model_pipeline.fit(X_train, y_train)

print("Linear Regression Model Trained Successfully!")
```

```python
# Predict house prices using testing data

y_pred = model_pipeline.predict(X_test)

# Display actual and predicted prices

comparison = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': y_pred
})

print(comparison.head(10))
```

```python
# Import evaluation metrics

import numpy as np
from sklearn.metrics import mean_squared_error, r2_score

# Calculate RMSE

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# Calculate R2 Score

r2 = r2_score(y_test, y_pred)

# Display results

print("Model Evaluation Results")
print("------------------------")
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))
print("R2 Score (%):", round(r2 * 100, 2), "%")
```

```python
# Get trained Linear Regression model

linear_model = model_pipeline.named_steps['model']

# Get feature names from preprocessing

feature_names = model_pipeline.named_steps[
    'preprocessor'
].get_feature_names_out()

# Create coefficient table

coefficients = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': linear_model.coef_
})

# Sort by coefficient value

coefficients = coefficients.sort_values(
    by='Coefficient',
    ascending=False
)

print("Linear Regression Coefficients")
print(coefficients.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(
    coefficients['Feature'],
    coefficients['Coefficient']
)

plt.xlabel("Coefficient Value")
plt.ylabel("Features")
plt.title("Feature Impact on House Prices")
plt.axvline(x=0, color='black', linewidth=0.8)

plt.tight_layout()
plt.show()
```

**Hyperparameter tuning**

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

# Create Random Forest pipeline

rf_tuning_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(random_state=42))
])

# Define parameters to test

param_grid = {
    'model__n_estimators': [100, 200, 300],
    'model__max_depth': [None, 5, 10, 15],
    'model__min_samples_split': [2, 5, 10],
    'model__min_samples_leaf': [1, 2, 4]
}

# Configure Grid Search

grid_search = GridSearchCV(
    estimator=rf_tuning_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1,
    verbose=1
)

# Train and find the best settings

grid_search.fit(X_train, y_train)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation RMSE:")
print(round(-grid_search.best_score_, 2))
```

```python
# Get the best Random Forest model

best_rf_model = grid_search.best_estimator_

# Predict using the unseen test data

tuned_rf_predictions = best_rf_model.predict(X_test)

# Calculate evaluation metrics

tuned_rf_rmse = np.sqrt(
    mean_squared_error(y_test, tuned_rf_predictions)
)

tuned_rf_r2 = r2_score(y_test, tuned_rf_predictions)

tuned_rf_mae = mean_absolute_error(
    y_test, tuned_rf_predictions
)

# Display results

print("TUNED RANDOM FOREST RESULTS")
print("---------------------------")
print("RMSE:", round(tuned_rf_rmse, 2))
print("R2 Score:", round(tuned_rf_r2, 4))
print("R2 Percentage:", round(tuned_rf_r2 * 100, 2), "%")
print("MAE:", round(tuned_rf_mae, 2))
```

**Let's try feature engineering**

```python
# Create copies of training and testing datasets

X_train_fe = X_train.copy()
X_test_fe = X_test.copy()

# Feature engineering function

def create_features(data):
    data = data.copy()

    data['total_rooms'] = (
        data['bedrooms'] + data['bathrooms']
    )

    data['area_per_bedroom'] = (
        data['area'] / data['bedrooms'].replace(0, 1)
    )

    data['total_facilities'] = (
        data['bathrooms'] + data['parking']
    )

    return data

# Apply feature engineering

X_train_fe = create_features(X_train_fe)
X_test_fe = create_features(X_test_fe)

print("Original feature count:", X_train.shape[1])
print("New feature count:", X_train_fe.shape[1])

print("\nNew features:")
print(X_train_fe.columns.tolist())
```

```python
# Create new numerical feature list

numerical_features_fe = [
    'area',
    'bedrooms',
    'bathrooms',
    'stories',
    'parking',
    'total_rooms',
    'area_per_bedroom',
    'total_facilities'
]

# Create preprocessing pipeline

preprocessor_fe = ColumnTransformer(
    transformers=[
        ('numerical', 'passthrough', numerical_features_fe),
        ('nominal',
         OneHotEncoder(drop='first', handle_unknown='ignore'),
         nominal_features),
        ('ordinal',
         OrdinalEncoder(categories=furnishing_order),
         ordinal_features)
    ]
)

# Create Linear Regression pipeline

lr_fe_pipeline = Pipeline([
    ('preprocessor', preprocessor_fe),
    ('model', LinearRegression())
])

# Train model

lr_fe_pipeline.fit(X_train_fe, y_train)

# Predict test data

lr_fe_predictions = lr_fe_pipeline.predict(X_test_fe)

# Evaluate

lr_fe_rmse = np.sqrt(
    mean_squared_error(y_test, lr_fe_predictions)
)

lr_fe_r2 = r2_score(y_test, lr_fe_predictions)

print("FEATURE ENGINEERING RESULTS")
print("---------------------------")
print("RMSE:", round(lr_fe_rmse, 2))
print("R2 Score:", round(lr_fe_r2, 4))
print("R2 Percentage:", round(lr_fe_r2 * 100, 2), "%")
```

**Validate using cross-validation**

```python
from sklearn.model_selection import cross_val_score
import numpy as np

# Original Linear Regression
original_cv_scores = cross_val_score(
    model_pipeline,
    X_train,
    y_train,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)

# Feature-engineered Linear Regression
engineered_cv_scores = cross_val_score(
    lr_fe_pipeline,
    X_train_fe,
    y_train,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)

# Convert scores to positive RMSE
original_cv_rmse = -original_cv_scores
engineered_cv_rmse = -engineered_cv_scores

print("5-FOLD CROSS-VALIDATION RESULTS")
print("--------------------------------")

print("\nOriginal Linear Regression")
print("Mean RMSE:", round(original_cv_rmse.mean(), 2))
print("Std RMSE:", round(original_cv_rmse.std(), 2))

print("\nFeature-Engineered Linear Regression")
print("Mean RMSE:", round(engineered_cv_rmse.mean(), 2))
print("Std RMSE:", round(engineered_cv_rmse.std(), 2))
```
